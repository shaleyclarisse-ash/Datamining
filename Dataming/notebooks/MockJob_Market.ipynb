```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../data/MockJob_Market.csv")

df.head(5)
```

```python
df.info()
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("\nDuplicated Values:")
print(df.duplicated().sum())
```

```python
print("Rows with Missing Values")
display(df[df.isnull().any(axis=1)])
```

```python
columns = [
    "Region"
    "Industry_Sector"
]
```

```python
columns = [
    "Region",
    "Industry_Sector",
    "Job_Title",
    "Experience_Level",
    "Education_Requirement",
    "Employment_Type",
    "Work_Mode",
    "Remote_Eligible",
    "Required_Skills",
    "Company_Size",
    "Posting_Source"
]

for col in columns:
    print(f"\n{col}")
    print(df[col].value_counts(dropna=False))
```

```python
df["Region"] = df["Region"].replace({
    "ncr" : "NCR"
})

print("NCR:",
      (df["Region"] == "NCR").sum())
```

```python
df["Industry_Sector"] = df["Industry_Sector"].replace({
    "IT" : "Information Technology"
})

print("Information Technology:",
      (df["Industry_Sector"] == "Information Technology").sum())
```

```python
df["Industry_Sector"] = df["Industry_Sector"].replace({
    "healthcare" : "Healthcare"
})

print("Healthcare:",
      (df["Industry_Sector"] == "Healthcare").sum())
```

```python
df["Region"] = df["Region"].replace({
    "Davao" : "Davao Region"
})

print("Region:",
      (df["Region"] == "Davao Region").sum())
```

```python
df["Employment_Type"] = df["Employment_Type"].replace({
    "Full time" : "Full-time"
})

print("Full-time:",
      (df["Employment_Type"] == "Full-time").sum())
```

```python
print("\nMissing Values")

missing = df.isnull().sum()

print(missing[missing>0])
```

```python
df["Region"] = df["Region"].fillna("Unknown")
df["Education_Requirement"] = df["Education_Requirement"].fillna("Not Specified")
df["Required_Skills"] = df["Required_Skills"].fillna("Not Specified")
```

```python
group_median = df.groupby(
    ["Job_Title", "Experience_Level"]
)

["Monthly_Salary_PHP"].transform("median")

df["Monthly_Salary_PHP"] = df["Monthly_Salary_PHP"].fillna(group_median)
```

```python
print(df.isnull().sum())
```

```python
print(df["Applicants_Count"].isnull().sum())
```

```python
group_median = df.groupby(
    ["Job_Title", "Experience_Level"]
)["Applicants_Count"].transform("median")

df["Applicants_Count"] = df["Applicants_Count"].fillna(
    group_median)
```

```python
sector_counts = df["Industry_Sector"].value_counts()

print("JOB POSTINGS BY INDUSTRY SECTOR")
print(sector_counts)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df["Posting_Date"] = pd.to_datetime(
    df["Posting_Date"],
    errors="coerce"
)

monthly_counts = (
    df.groupby(
        df["Posting_Date"].dt.to_period("M")
    )
    .size()
    .sort_index()
)

print("JOB POSTINGS BY MONTH")
print(monthly_counts)
```

```python
plt.figure(figsize=(11, 5))

plt.plot(
    monthly_counts.index.astype(str),
    monthly_counts.values,
    color="#800000",
    marker="o",
    linewidth=2.5,
    markersize=8
)

plt.title(
    "Monthly Job Posting Trends (2025)",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Posting Month")
plt.ylabel("Number of Job Postings")

# Display exact values on each point
for i, value in enumerate(monthly_counts.values):
    plt.annotate(
        str(value),
        (i, value),
        textcoords="offset points",
        xytext=(0, 10),
        ha="center"
    )

plt.xticks(rotation=45)
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
salary_sector = (
    df.groupby("Industry_Sector")["Monthly_Salary_PHP"]
    .agg(["mean", "median"])
    .sort_values("median")
)

print("MEAN AND MEDIAN SALARY BY INDUSTRY")
print(salary_sector.round(2))

ax = salary_sector.plot(
    kind="barh",
    figsize=(11, 6),
    color=["#800000", "#D99AA5"],
    edgecolor="black"
)

plt.title("Mean vs Median Monthly Salary by Industry Sector",
          fontsize=14, fontweight="bold")
plt.xlabel("Monthly Salary (PHP)")
plt.ylabel("Industry Sector")
plt.legend(["Mean", "Median"])

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

experience_order = ["10+ years", 
                    "4–6 years",
"Entry-level",
"7–10 years",
"1–3 years"]

salary_experience = (
    df.groupby("Experience_Level")["Monthly_Salary_PHP"]
    .agg(["mean", "median"])
    .reindex(experience_order)
)

print("MEAN AND MEDIAN SALARY BY EXPERIENCE LEVEL")
print(salary_experience.round(2))

ax = salary_experience.plot(
    kind="bar",
    figsize=(10, 6),
    color=["#800000", "#D99AA5"],
    edgecolor="black",
    width=0.65
)

plt.title(
    "Mean vs Median Monthly Salary by Experience Level",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Experience Level")
plt.ylabel("Monthly Salary (PHP)")
plt.xticks(rotation=0)
plt.legend(["Mean", "Median"])

# Add salary labels
for container in ax.containers:
    ax.bar_label(
        container,
        fmt="₱%.0f",
        padding=3,
        fontsize=9
    )

plt.tight_layout()
plt.show()
```
