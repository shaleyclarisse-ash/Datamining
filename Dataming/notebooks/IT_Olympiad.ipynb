```python
from pathlib import Path
import pandas as pd

# Get the current working directory
print("Current directory:", Path.cwd())

# Locate the data folder
data_path = Path.cwd() / "data"

if not data_path.exists():
    data_path = Path.cwd().parent / "data"

print("Data folder:", data_path.resolve())
print("Folder exists:", data_path.exists())
```

```python
df_2021 = pd.read_excel(data_path / "Jobposting_2021_Datasets.xlsx")
df_2022 = pd.read_excel(data_path / "Jobposting_2022_Datasets.xlsx")
df_2023 = pd.read_excel(data_path / "Jobposting_2023_Datasets.xlsx")
df_2024 = pd.read_excel(data_path / "Jobposting_2024_Datasets.xlsx")
df_2025 = pd.read_excel(data_path / "Jobposting_2025_Datasets.xlsx")

datasets = {
    2021: df_2021,
    2022: df_2022,
    2023: df_2023,
    2024: df_2024,
    2025: df_2025
}

print("All datasets loaded successfully!")
```

```python
for year, df in datasets.items():
    print(f"\n===== {year} =====")

    if year == 2021:
        date_column = "created_at"
    elif year == 2023:
        date_column = "Posting_Date"
    elif year == 2024:
        date_column = "posting_date"
    elif year == 2025:
        date_column = "Date Posted"
    else:
        date_column = "Posting Date"

    dates = pd.to_datetime(
        df[date_column],
        errors="coerce"
    )

    print(dates.dt.year.value_counts(dropna=False))
```

```python
column_mapping = {
    2021: {
        "title": "Job_Title",
        "skills_required": "Required_Skills",
        "created_at": "Posting_Date",
        "location": "Job_Location"
    },
    2022: {
        "Job Title": "Job_Title",
        "Required Skills": "Required_Skills",
        "Posting Date": "Posting_Date",
        "Job Location": "Job_Location"
    },
    2023: {
        "Job Title": "Job_Title",
        "Required Skills": "Required_Skills",
        "Posting_Date": "Posting_Date",
        "Job Location": "Job_Location"
    },
    2024: {
        "job_title": "Job_Title",
        "required_skills": "Required_Skills",
        "posting_date": "Posting_Date",
        "job_location": "Job_Location"
    },
    2025: {
        "Job Title": "Job_Title",
        "Skills Required": "Required_Skills",
        "Date Posted": "Posting_Date",
        "City": "Job_Location"
    }
}

standardized = []

for year, df in datasets.items():

    temp = df.rename(columns=column_mapping[year]).copy()

    temp = temp[
        ["Job_Title", "Required_Skills",
         "Posting_Date", "Job_Location"]
    ]

    temp["Source_File_Year"] = year

    standardized.append(temp)

print("Column names standardized!")
```

```python
df_combined = pd.concat(
    standardized,
    ignore_index=True
)

print("Combined dataset shape:", df_combined.shape)

display(df_combined.head())
```

```python
df_clean = df_combined.copy()

df_clean["Posting_Date"] = pd.to_datetime(
    df_clean["Posting_Date"],
    errors="coerce"
)

df_clean["Year"] = df_clean["Posting_Date"].dt.year

# Remove exact duplicate records
df_clean = df_clean.drop_duplicates()

# Retain only the competition years
df_clean = df_clean[
    df_clean["Year"].between(2021, 2025)
].copy()

# Remove records without usable skills
df_clean = df_clean.dropna(subset=["Required_Skills"])

df_clean = df_clean[
    df_clean["Required_Skills"].astype(str).str.strip() != ""
]

print("Cleaned dataset shape:", df_clean.shape)
print(df_clean["Year"].value_counts().sort_index())
```

```python
df_skills = df_clean.copy()

df_skills["Required_Skills"] = (
    df_skills["Required_Skills"]
    .astype(str)
    .str.split(",")
)

df_skills = df_skills.explode("Required_Skills")

df_skills["Skill"] = (
    df_skills["Required_Skills"]
    .str.strip()
    .str.lower()
)

display(df_skills.head())
```

```python
skill_mapping = {
    "py": "python",
    "python programming": "python",
    "structured query language": "sql",
    "ms excel": "excel",
    "microsoft excel": "excel",
    "powerbi": "power bi",
    "power bi desktop": "power bi",
    "ml": "machine learning",
    "artificial intelligence": "ai",
    "javascript": "javascript",
    "js": "javascript"
}

df_skills["Skill"] = (
    df_skills["Skill"]
    .replace(skill_mapping)
)

df_skills = df_skills[
    df_skills["Skill"].notna()
    & (df_skills["Skill"] != "")
]

print("Skill extraction completed!")
```

```python
yearly_totals = df_clean.groupby("Year").size()

print(yearly_totals)
```

```python
df_clean = df_clean.reset_index(drop=True)
df_clean["Posting_ID"] = df_clean.index

df_skills = df_clean.copy()

df_skills["Skill"] = (
    df_skills["Required_Skills"]
    .astype(str)
    .str.split(",")
)

df_skills = df_skills.explode("Skill")

df_skills["Skill"] = (
    df_skills["Skill"]
    .str.strip()
    .str.lower()
    .replace(skill_mapping)
)

df_skills = df_skills[
    df_skills["Skill"].notna()
    & (df_skills["Skill"] != "")
]

# Count each skill once per job posting
df_skills = df_skills.drop_duplicates(
    subset=["Posting_ID", "Skill"]
)

skill_counts = (
    df_skills.groupby(["Year", "Skill"])
    ["Posting_ID"]
    .nunique()
    .reset_index(name="Job_Count")
)

skill_counts["Total_Postings"] = (
    skill_counts["Year"].map(yearly_totals)
)

skill_counts["Skill_Share"] = (
    skill_counts["Job_Count"]
    / skill_counts["Total_Postings"] * 100
)

display(skill_counts.head())
```
