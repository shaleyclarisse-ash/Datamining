```python

sector_counts = df["title"].value_counts().head(10)

print("Title")
print(sector_counts)

plt.figure(figsize=(10, 6))

sector_counts.sort_values().plot(
    kind="barh",
    color="maroon",
    )


plt.title("Job Postings by Title")
plt.xlabel("Number of Job Postings")
plt.ylabel("Title")

plt.tight_layout()
plt.show()
```

```python
sector_counts = df["skills_required"].value_counts().tail(10)

print("Skills Required")
print(sector_counts)

plt.figure(figsize=(10, 9))

sector_counts.sort_values().plot(
    kind="barh",
    color="maroon",
    )


plt.title("Top 10 Emerging Skills 2021")
plt.xlabel("Number of Job Postings")
plt.ylabel("Required Skills")

plt.tight_layout()
plt.xticks(rotation=45)
plt.show()
```
