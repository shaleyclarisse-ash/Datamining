```python
df = df.drop_duplicates()
```

```python
print("\nMissing Values: ")
print(df.isnull().sum())
```

```python
print("Rows with Missing Values:")
display(df[df.isnull().any(axis=1)])
```

```python
df = df.fillna({"Month" : "December",
                "Year" : "2021",
                "Day" : "5",
                "Product_Category" : "Bikes",
                "Age_Group" : "Adults (35-64)",
                "Sub_Category" : "Mountain Bikes"})
```

```python
df.loc[21, "Product_Description"] = "Mountain-200 Black, 42"
df.loc[22, "Order_Quantity"] = 1
```

```python
df.columns = df.columns.str.strip()

money_cols = [
    'Unit_Cost',
    'Unit_Price',
    'Cost',
    'Revenue',
    'Profit'
]

for col in money_cols:
    df[col] = pd.to_numeric(
        df[col].astype(str).str.replace(r'[$,]', '', regex=True),
        errors='coerce'
    )

df['Order_Quantity'] = pd.to_numeric(
    df['Order_Quantity'], errors='coerce'
)

print(df[money_cols].isna().sum())
```

```python
print(df['Customer_Gender'].value_counts(dropna=False))
```

```python
df['Customer_Gender'] = (
    df['Customer_Gender']
    .where(df['Customer_Gender'].isin(['F', 'M']), 'Unknown')
)

print(df['Customer_Gender'].value_counts(dropna=False))
```
