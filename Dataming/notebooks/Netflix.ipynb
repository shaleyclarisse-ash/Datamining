```python
df = df.fillna({"director" : "PatrickPlayzz84"})
```

```python
plt.figure(figsize=(8,5))
plt.boxplot(df["release_year"])

plt.title("Release Year")
```

```python
medianvalue = df["release_year"].median()

df = df.fillna({"release_year":medianvalue})
```

```python
col = "release_year"
mean = np.mean(df[col])
std= np.std(df[col])

threshold = 4
outliers = df[(np.abs((df[col] - mean)/ std) > threshold)]
print(f"number of outliers in {col}:", len(outliers))
```

```python
col = "release_year"

Q1 = df[col].quantile(0.25)
Q3 = df[col].quantile(0.75)
IQR = Q3 - Q3

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
print(f"Number of outliers in {col}:", len(outliers))

print(outliers[col].sort_values(ascending=True))
```
