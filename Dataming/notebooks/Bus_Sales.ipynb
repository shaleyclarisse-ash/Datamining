```python
import pandas as pd

df = pd.read_csv("../data/Bus_Sales.csv")

raw_df = df.copy()
clean_df = df.copy()

print("Original dataset:", raw_df.shape)
print("Working dataset:", clean_df.shape)

duplicates = clean_df[
    clean_df.duplicated(
        subset=["Transaction_ID"],
        keep=False
    )
]

print("Duplicate ID records:")
print(duplicates.to_string(index=False))
```

```python
print("Exact duplicate rows:")
print(clean_df.duplicated().sum())
```

```python

clean_df = clean_df.drop_duplicates()

print("Remaining rows:", len(clean_df))
```

```python
columns = [
    "Branch",
    "Sales_Channel",
    "Product",
    "Category",
    "Payment_Method",
    "Promotion"
]

for col in columns:
    print(f"\n{col}")
    print(clean_df[col].value_counts(dropna=False))
```

```python
df_clean = df.copy()

df_clean["Branch"] = df_clean["Branch"].replace({
    "digos city" : "Digos City"
})

print(df_clean["Branch"].value_counts())
```

```python
invalid_qty = clean_df[
    (clean_df["Quantity"] <= 0) |
    (clean_df["Quantity"].isna())
]

print("INVALID QUANTITY RECORDS")
print(invalid_qty.to_string(index=False))

print("\nTotal invalid quantities:", len(invalid_qty))
```

```python
print(clean_df.loc[
    clean_df["Quantity"] == -2,
    [
        "Transaction_ID",
        "Product",
        "Quantity",
        "Unit_Price_PHP",
        "Gross_Sales_PHP",
        "Discount_PHP",
        "Net_Sales_PHP"
    ]
].to_string(index=False))
```

```python
mask = clean_df["Quantity"] == -2

print("Matching records:", mask.sum())
```

```python
clean_df.loc[mask, "Quantity"] = 3

print("Quantity corrected successfully!")
```

```python
clean_df.loc[mask, "Gross_Sales_PHP"] = (
    clean_df.loc[mask, "Quantity"] *
    clean_df.loc[mask, "Unit_Price_PHP"]
)

clean_df.loc[mask, "Net_Sales_PHP"] = (
    clean_df.loc[mask, "Gross_Sales_PHP"] -
    clean_df.loc[mask, "Discount_PHP"]
)

print("Sales values recalculated!")
```

```python
print("Remaining invalid quantities:")
print((clean_df["Quantity"] <= 0).sum())

print("\nCorrected transaction:")
print(clean_df.loc[
    mask,
    [
        "Transaction_ID",
        "Product",
        "Quantity",
        "Gross_Sales_PHP",
        "Net_Sales_PHP"
    ]
])
```
